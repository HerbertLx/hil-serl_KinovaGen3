```python
#!/usr/bin/env python3
import rclpy
from rclpy.node import Node
from tf2_ros import Buffer, TransformListener
from geometry_msgs.msg import TransformStamped
import time

class EndEffectorPoseReader(Node):
    def __init__(self):
        super().__init__('end_effector_pose_reader')
        self.tf_buffer = Buffer()
        self.tf_listener = TransformListener(self.tf_buffer, self)

        # 等待 TF 树建立（可选：加一点延迟）
        time.sleep(1.0)

        self.get_current_pose()

    def get_current_pose(self):
        try:
            # 查询 base_link -> end_effector_link 的变换
            transform: TransformStamped = self.tf_buffer.lookup_transform(
                target_frame='base_link',
                source_frame='end_effector_link',
                time=rclpy.time.Time(),  # 最新可用
                timeout=rclpy.duration.Duration(seconds=1.0)
            )

            trans = transform.transform.translation
            rot = transform.transform.rotation

            self.get_logger().info(
                f"当前末端执行器位姿（相对于 base_link）:\n"
                f"  位置 (x, y, z): [{trans.x:.4f}, {trans.y:.4f}, {trans.z:.4f}] 米\n"
                f"  四元数 (x, y, z, w): [{rot.x:.4f}, {rot.y:.4f}, {rot.z:.4f}, {rot.w:.4f}]"
            )

        except Exception as e:
            self.get_logger().error(f"无法获取 TF 变换: {e}")

def main(args=None):
    rclpy.init(args=args)
    node = EndEffectorPoseReader()
    rclpy.shutdown()

if __name__ == '__main__':
    main()
```
